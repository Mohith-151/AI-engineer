### How to call an API by requests, and all about API

```python
import requests
url = "https://jsonplaceholder.typicode.com/users/1"
response  = requests.get(url , timeout=10)
print("connected successfull -----status code", response.status_code)
```

```python
print(type(response))
```

```python
data = response.json()
print(type(data))
```

```python
print(data)
```

```python
import pandas as pd
df = pd.DataFrame([data])
```

```python
df
```

```python
from tabulate import tabulate

print(tabulate(data.items(), headers=["Field", "Value"], tablefmt="grid"))
```

```python
posts = requests.get(
    "https://jsonplaceholder.typicode.com/posts",
    params={"userId": 1},   # becomes ?userId=1
    timeout=10,
).json()

for post in posts:
    print(post["id"], post["title"])
```
