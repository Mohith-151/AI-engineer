***Class:*** is a blueprint which contains data, funtions.\
***Object*** is a thing which is built using the class.\

***__init__*** in a defualt funtion of a class which will be called automatically when object is created.\
***self*** is a convention (not a keyword) which is used on variable, functions and all which makes those are only relavent to that particular class/object.

- but.., def move(self): &#10142; works perfectlt

***HTTP*** A request-respond protocol.\
***GET*** a requesting method for any resources which does not rewrites or messes with server's data.\
***Status code*** is a 3-digit code comes along with the response.\
***JSON*** a programming language used to send data over HTTP to server.

#### Status codes


| Range	| Meaning	|Examples|
| --- | --- | --- |
| 2xx	| Success	| 200 OK, 201 Created, 204 No Content|
| 3xx	| Redirect|	301 Moved Permanently, 302 Found|
| 4xx	| Your mistake	| 400 Bad Request, 401 Unauthorized (bad or missing credentials), 403 Forbidden, 404 Not Found, 429 Too Many Requests(rate limited)|
| 5xx	| Server's problem	| 500 Internal Server Error, 503 Service Unavailable|

A sample request code..

```python
import requests
response = requests.get("https://jsonplaceholder.typicode.com/todos/1")

print(response.status_code)  # 200
print(response.json())       # parsed dict
```
