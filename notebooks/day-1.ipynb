````python
def count_words(text): 
  word_list= {}
  for word in text.split():
    word_list[word] = word_list.get(word, 0)+1
  return word_list

text = 'The oldest trees in the world were already ancient seedlings when the Egyptian pyramids were being constructed. Deep beneath our feet, vast networks of fungal threads share nutrients and emergency warnings across entire forests.'
output = count_words(text)
# print(output)
for key, value in output.items():
  print(f'{key} : {value}')
````

##### Data structure
**LIST :** 
- A linear data structure where the elements can be accesed using the index numbers, 
- and the Items it holds can be any datatype and all will be on order. 
- ANd the items in List are Mutable we can change the values of the item anytime
- Example: [[1,2,3], 'data sciecne', 'A001']

**TUPLE :** 
- Also a Linear data structure, elements accesed usinf Index number. 
- But the items in the Tuple in not Mutable, once created we can't change the value of that items.
- Does not Allow Duplicate Values
- Items will be Order
- Exmaple: ('Apple', 1001, "A0051")

**Dict :**
- Dictionary ,Key-value form
- NON-linear data structure, where elements will not in the order when print multiple times
- NO duplicate key is allowed, each key should be unique as the item is accesed using this key not my index numbers
- exmaple: ageOfPerson = {'Mohith': 21, 'Tejas': 20, 'Dinesh': 24}

**SET :**
- A non-Linear data struture, where items are not in order.
- the items are immutable, as the item order changes everytime the code runs it is immposible to located the particular value and change it, best wat is to remove that item and add new one.
- exmaple: {'banana', 23, 'mango', 'AF421'}


##### GIT push and commit

***GIT-HUB*** is like a cloud database exclusivly for codes and projects.\
where as ***git*** is a wireless connection between our computer and the github database, used to tranfers files and all.

**Google version defination:**
- ***Git*** is a Time Machine for your computer. It is a tool (a Version Control System) that you install locally. \
It watches your project folder and takes "snapshots" of your code every time you save a milestone.\
If you make a mistake, Git lets you roll back the clock to an older version. It works entirely offline.

- ***GitHub*** is a Social Cloud Platform. \
It is a website where you can upload copies of those Git snapshots. It acts as a remote backup, \
allows you to share your project with the world, and lets multiple developers work on the same codebase simultaneously.

##### git commands to make repo from the computer command line

- step 1: open the folder in the CMD
- step 2: inital git by ```git init ```

````python
# 10. Count a class label without count()
# Given a list of class labels and a target label, count how often the target appears.
# labels = [0, 1, 1, 2, 0, 1, 2, 1], target 1 → 4

labels = [0, 1, 1, 2, 0, 1, 2, 1]
count =0
for num in labels:
  if num == 1:
    count+=1
print(count)
````

````python
# 13. Fit two documents into a context window
# Given a list of document token counts and a token budget, find all pairs that add up exactly to the budget. 
#Each pair should appear once. [1200, 2800, 1500, 2500, 3000, 1096], budget 4000 → [(1200, 2800), (1500, 2500)]

token_counts = [1200, 2800, 1500, 2500, 3000, 1096]
token_budget = 4000
budget_pair =[]
# for i in range(len(token_counts)):
#   for j in range(i+1,len(token_counts)):
#     if token_counts[i]+token_counts[j] == 4000:
#      pair =sorted([token_counts[i],token_counts[j]])
#      if pair not in budget_pair:
#        budget_pair.append(pair)
# print(budget_pair) #slower

seen = set()
for count in token_counts:
  comp = token_budget - count
  if comp in seen:
    pair = (min(count,comp), max(count,comp))
    budget_pair.append(pair)
  seen.add(count)
print(budget_pair)
````

````python
# 14. Clean a noisy dataset row
# Given a raw list mixing numbers, strings, None, and nested lists, keep only the numeric values.
# [3.5, "cat", 7, [1, 2], "10", 0.25, None] → [3.5, 7, 0.25]
# Watch out: True and False count as int in Python. Decide whether to exclude them.
# Bonus: also pull out the numbers inside nested lists.

dataset = [3.5, "cat", 7, [1, 2], "10", 0.25, None, True]
# num_dataset=[]
# for item in dataset:
#   if item.dtype == int:
#     num_dataset.append(item)
# print(num_dataset) #it doesn't handles float dtype

def clean_data(dataset):
  num_dataset = []
  for item in dataset:
    if isinstance(item,list):
      num_dataset.extend(clean_data(item))
    elif isinstance(item, (int,float)) and not isinstance(item , bool):
      num_dataset.append(item)
  return num_dataset

result = clean_data(dataset)
print(result)
````

````python
# 15. Longest improving streak in validation accuracy
# Given accuracy per epoch, return the longest run of strictly increasing values.
# [0.62, 0.65, 0.64, 0.68, 0.71, 0.73, 0.70] → [0.64, 0.68, 0.71, 0.73]

epoch_set = [0.62, 0.65, 0.64, 0.68, 0.71, 0.73, 0.70]

current = [epoch_set[0]]   # start with the first number
best = []

for i in range(1, len(epoch_set)):
    if epoch_set[i] > epoch_set[i-1]:
        current.append(epoch_set[i])     # streak continues
    else:
        current = [epoch_set[i]]         # streak broke, start over

    if len(current) > len(best):
        best = current[:]                # save a copy of the longest so far

print(best)
````
