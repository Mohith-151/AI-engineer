**The right Code is...**

````python
def count_words(text): 
  word_list= {}
  for word in text.split():
    word_list[word] = word_list.get(word, 0)+1
  return word_list

text = 'The oldest trees in the world were already ancient seedlings when the Egyptian pyramids were being constructed. Deep beneath our feet, vast networks of fungal threads share nutrients and emergency warnings across entire forests.'
output = count_words(text)
# print(output)
for key, value in output.items():
  print(f'{key} : {value}')
````

##### Data structure
**LIST :** 
- A linear data structure where the elements can be accesed using the index numbers, 
- and the Items it holds can be any datatype and all will be on order. 
- ANd the items in List are Mutable we can change the values of the item anytime
- Example: [[1,2,3], 'data sciecne', 'A001']

**TUPLE :** 
- Also a Linear data structure, elements accesed usinf Index number. 
- But the items in the Tuple in not Mutable, once created we can't change the value of that items.
- Does not Allow Duplicate Values
- Items will be Order
- Exmaple: ('Apple', 1001, "A0051")

**Dict :**
- Dictionary ,Key-value form
- NON-linear data structure, where elements will not in the order when print multiple times
- NO duplicate key is allowed, each key should be unique as the item is accesed using this key not my index numbers
- exmaple: ageOfPerson = {'Mohith': 21, 'Tejas': 20, 'Dinesh': 24}

**SET :**
- A non-Linear data struture, where items are not in order.
- the items are immutable, as the item order changes everytime the code runs it is immposible to located the particular value and change it, best wat is to remove that item and add new one.
- exmaple: {'banana', 23, 'mango', 'AF421'}


##### GIT push and commit

***GIT-HUB*** is like a cloud database exclusivly for codes and projects.\
where as ***git*** is a wireless connection between our computer and the github database, used to tranfers files and all.

**Google version defination:**
- ***Git*** is a Time Machine for your computer. It is a tool (a Version Control System) that you install locally. \
It watches your project folder and takes "snapshots" of your code every time you save a milestone.\
If you make a mistake, Git lets you roll back the clock to an older version. It works entirely offline.

- ***GitHub*** is a Social Cloud Platform. \
It is a website where you can upload copies of those Git snapshots. It acts as a remote backup, \
allows you to share your project with the world, and lets multiple developers work on the same codebase simultaneously.

##### git commands to make repo from the computer command line

- step 1: open the folder in the CMD
- step 2: inital git by ``` init git ```
- step 3: add the files&folder . by - ```git add .```
- step 4: firat commit, by ```git commit -m "inital commit"``` , -m means main branch
- step 5: go to gihub website and create a repo, avoid click add README file if u already have a README file in local folder
- copy the repo URL
- step 6: paste the URL after ```git remote add origin PASTE URL``` 
- step 7: setting MAIN branch. ```git  branch -M main```
- step 8: push to github. ```git push -u origin main```
