```python
# 1. word_frequency(tokens)
# Return a dict with the count of each word.
# ["the","cat","the","dog"] → {"the": 2, "cat": 1, "dog": 1}

tokens= ["the","cat","the","dog"]

def word_frequency(tokens):
  freq_dict= dict()
  for word in tokens:
    if word not in freq_dict:
      freq_dict[word]=1
    else:
      freq_dict[word]+=1
  return freq_dict

result = word_frequency(tokens)
print(result)
```

```python
# 2. build_vocab(tokens)
# Give each unique word an ID, starting from 0, in order of first appearance.
# ["the","cat","the","dog"] → {"the": 0, "cat": 1, "dog": 2}

tokens = ["the","cat","the","dog"]
def build_vocab(tokens):
  words_ID = dict()
  for word in tokens:
    if word not in words_ID:
      words_ID[word] = len(words_ID)
  return words_ID

vocab = build_vocab(tokens)
print(vocab)
```

```python
# 3. encode(tokens, vocab)
# Convert words to IDs using a vocab. Unknown words become -1.
# ["the","dog","fish"] with the vocab from #2 → [0, 2, -1]

vocab = {'the': 0, 'cat': 1, 'dog': 2}
new_token = ["the","dog","fish"]

def encode(tokens, vocab):
  id_list =[]
  for token in tokens:
    if token in vocab:
      id_list.append(vocab[token])
    else:
      id_list.append(-1)
  return id_list

result = encode(new_token, vocab)
print(result)
```

```python
# 4. invert_dict(vocab)
# Swap keys and values so you can go from ID back to word.
# {"the": 0, "cat": 1} → {0: "the", 1: "cat"}

normal_dict = {"the": 0, "cat": 1}
def invert_dict(vocab):
  rev_dict = dict()
  for key, values in vocab.items():
    rev_dict[values]= key
  return rev_dict

result = invert_dict(normal_dict)
print(result)
```

```python
# 7. top_k(scores, k)
# Return the k keys with the highest scores, best first.
# {"cat": 0.7, "dog": 0.2, "bird": 0.1}, k=2 → ["cat", "dog"]

scores={"cat": 0.7, "dog": 0.2, "bird": 0.1}
k=2

def top_k(scores,k):
  sorted_list = sorted(scores, key=scores.get, reverse=True)
  return sorted_list[:k]

result= top_k(scores,k)
print(result)
```

```python
# 9. accuracy_per_class(true, pred)
# For each true label, calculate the fraction of predictions that were correct.
# true=["a","a","b","b"], pred=["a","b","b","b"] → {"a": 0.5, "b": 1.0}

#I took help of AI cuz i didn't understood the questions
true=["a","a","b","b"]
pred=["a","b","b","b"]

def accuracy_per_class(true, pred):
  total ={}
  correct ={}
  for t,p in zip(true, pred):
    total[t]= total.get(t,0)+1
    if t == p:
      correct[t]= correct.get(t,0)+1

  score ={}
  for item in total:
    score[item] = correct.get(item, 0)/total.get(item)
  return score

acc_score = accuracy_per_class(true, pred)
print(acc_score)
```
